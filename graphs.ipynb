```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
import yfinance as yf
import pandas as pd
import matplotlib.dates as mdates
```

```python
df = yf.download("WIPRO.NS", start="2010-01-01", end=pd.Timestamp.today().strftime("%Y-%m-%d"))
```

```python
print(df.head())
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df['Close'], color='blue')
plt.title('WIPRO Stock Closing Price History')
plt.xlabel('Date')
plt.ylabel('Close Price (INR)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print(df[['Close']])
```

```python
df_last_month = df.tail(30)
```

```python
plt.figure(figsize=(10,5))
plt.plot(df_last_month.index, df_last_month['Close'], marker='o', label='WIPRO Closing Price')
plt.title("WIPRO Closing Price - Last 30 Days")
plt.xlabel("Date")
plt.ylabel("Close Price (INR)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
